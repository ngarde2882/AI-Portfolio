```python
import matplotlib.pyplot as plt
import numpy as np
```

```python

# cols [trials,threads,pi,error,time]
data = open('part1.out','r')
D = []
for line in data:
    L = line.split(',')
    for i in range(len(L)):
        L[i] = float(L[i].split(' ')[-1])
    D.append(L)
labels = ['trials','threads','pi','relative error','time (s)']
print(labels)
for i in D:
    print(i)
D = np.array(D)
print(D)
```

```python
x,y1,y2,y3 = [],[],[],[]
T1 = D[0][4]
for row in D[:14]:
    x.append(row[1])
    y1.append(row[4])
    S = T1/row[4]
    y2.append(S)
    y3.append(S/row[1])

fig, ax = plt.subplots(1,3, figsize=(14, 3))
ax[0].plot(x, y1)
ax[0].set_xscale('log')
ax[0].set_xlabel(labels[1])
ax[0].set_ylabel(labels[4])
ax[0].set_title('Execution Time x pThreads n=10^8')

ax[1].plot(x, y2)
ax[1].set_xscale('log')
ax[1].set_xlabel(labels[1])
ax[1].set_ylabel('Speedup Sp=T1/Tp')
ax[1].set_title('Execution Speedup x pThreads n=10^8')

ax[2].plot(x, y3)
ax[2].set_xscale('log')
ax[2].set_xlabel(labels[1])
ax[2].set_ylabel('Efficiency Ep=Sp/p')
ax[2].set_title('Execution Efficiency x pThreads n=10^8')
# ax.legend()
plt.show()

print(2**np.argmin(np.array(y1)))
```
